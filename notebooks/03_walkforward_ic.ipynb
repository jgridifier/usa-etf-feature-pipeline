# 03 Walk-forward IC + thematic rotation

Research only — not investment advice.

Demonstrates:
1. Walk-forward Spearman IC of composite S_i vs **next-month** excess (no same-month leakage).
2. ScoreSimple XSD rotation ON/OFF vs next-month excess (see xsd_rotation_memo.md).


```python
from pathlib import Path
import pandas as pd
from usa_etf_features.prices import load_adj_close_csv
from usa_etf_features.universe import load_universe_csv, category_map, load_universe_config, eligible_tickers
from usa_etf_features.walkforward import walkforward_ic_table, rotation_on_off_next_month_table, write_ic_csv

ROOT = Path("..")
UNI = ROOT / "data/raw/usa_universe_categorized.csv"
PX = Path("/workspace/investments/growth_alpha_adj_close.csv")
if not PX.exists():
    PX = ROOT / "data/raw/growth_alpha_adj_close_sample.csv"

uni = load_universe_csv(UNI)
cfg = load_universe_config()
cats = category_map(uni, cfg)
px = load_adj_close_csv(PX)
tickers = [t for t in ["VOO","QQQ","QQQM","IJR","XSD","XBI","QUAL","IWM"] if t in px.columns]
cat_s = pd.Series({t: cats.get(t, "Unknown") for t in tickers})
ic = walkforward_ic_table(px, cat_s, tickers=tickers, benchmark="VOO")
print(ic.tail())
print("mean IC", ic["IC"].mean())
write_ic_csv(ic, ROOT / "data/processed/ic_walkforward_notebook.csv")
rotation_on_off_next_month_table(px, ticker="XSD", use_vol_gate=True)
```
